```python
import numpy as np
import matplotlib.pyplot as plt

# 假设文件路径为'my_data.npy'
file_path = 'token_num_list.npy'

# 读取npy文件中的数据
data = np.load(file_path)

# 定义区间长度
bin_size = 16
bins = np.arange(0, 513, bin_size)

# 画柱状图
plt.hist(data, bins=bins, edgecolor='black')

# 设置图表标题和标签
plt.title("SMILES' Description tokens number distribution")
plt.xlabel('token num')
plt.ylabel('Frequency')

# 显示图表
plt.show()
```
